```python
import uproot
import awkward as ak
import pandas as pd

#library akward, bo czesc danych to wekrory dlugosic 1... - pd ma z tym problem
file = uproot.open(r"../../Data_Separated/Bias/MLDataBias.root")

ak_elec = file["MLDataElec;1"].arrays(library="ak")
ak_muon = file["MLDataMuon;1"].arrays(library="ak")

df_elec = ak.to_dataframe(ak_elec)
df_muon = ak.to_dataframe(ak_muon)

print("Electrons shape:", df_elec.shape)
print("Muons shape:    ", df_muon.shape)
print("Columns:", df_elec.columns.tolist())

df_elec.to_csv(r"../../Data_Separated/Bias/electronBias.csv", index=False)
df_muon.to_csv(r"../../Data_Separated/Bias/muonBias.csv", index=False)

print("Zapisano jako CSV!")
```
