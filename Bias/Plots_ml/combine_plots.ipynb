```python
import os
from PIL import Image

# folder z obrazami
folder = r"C:\Users\qmiko\Desktop\cern\Plots_ml"

# grupy metod
methods = ["Forest", "Knn", "Mlp"]

# dla każdej metody łączymy jej obrazy w jeden
for method in methods:
    # wybieramy pliki metody
    files = [f for f in os.listdir(folder) if f.startswith(method) and f.endswith(".png")]
    files.sort()  # opcjonalnie, żeby zawsze matrix pierwszy, roc drugi
    
    # wczytanie obrazów
    images = [Image.open(os.path.join(folder, f)) for f in files]
    
    # obliczenie wymiarów finalnego obrazu (poziomo)
    widths, heights = zip(*(i.size for i in images))
    total_width = sum(widths)
    max_height = max(heights)
    
    # stworzenie pustego obrazu
    new_im = Image.new('RGB', (total_width, max_height), color=(255,255,255))
    
    # wklejanie obrazów obok siebie
    x_offset = 0
    for im in images:
        new_im.paste(im, (x_offset,0))
        x_offset += im.size[0]
    
    # zapis nowego pliku
    output_path = os.path.join(folder, f"{method}_combined.png")
    new_im.save(output_path)
    print(f"Zapisano: {output_path}")
```
